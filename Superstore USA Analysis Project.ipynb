```python
def purva(a):
    cat = list(a.columns[a.dtypes == "object"])
    for col in a.columns:
        if col in cat:
            mode = a[col].mode()[0]
            a[col] = a[col].fillna(mode)  
        else:
            mean = a[col].mean()
            a[col] = a[col].fillna(mean)  
```

```python
# 1) What is the total profit per region?

top_seven = a.groupby("Region")["Profit"].sum().nlargest(7).index 
filtered_df = a[a["Region"].isin(top_seven)]
plt.figure(figsize = (12, 6))
sns.barplot(data = filtered_df, x = "Region", y = "Profit", order = top_seven, palette = "Set2", edgecolor = "black")
plt.title("Top 7 Regions by Total Profit - Profit Distribution", fontsize = 15)
plt.xlabel("Region", fontsize = 14)
plt.ylabel("Profit", fontsize = 14)
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# 2) Find the average discount given by product category and sub-category?

top_categories = a.groupby("Product Category")["Discount"].mean().nlargest(7).index
filtered_a = a[a["Product Category"].isin(top_categories)]
pivot_discount = filtered_a.pivot_table(
    values = "Discount",
    index = "Product Category",
    columns = "Product Sub-Category",
    aggfunc = "mean"
)
plt.figure(figsize = (12, 6))
sns.heatmap(pivot_discount, annot = True, fmt = ".2f", cmap = "YlGnBu", linewidth = 0.5)
plt.title("Average Discount By Top 7 Product Categories Sub-Categories", fontsize = 15)
plt.xlabel("Product Sub-Categories", fontsize = 14)
plt.ylabel("Product Categories", fontsize = 14)
plt.tight_layout()
plt.show()
```

```python
# 3) What is the total sales per customer segment and ship mode?

sales_grouped = a.groupby(["Customer Segment", "Ship Mode"])["Sales"].sum().reset_index()
plt.figure(figsize = (12, 6))
sns.barplot(data = sales_grouped, x = "Customer Segment", y = "Sales", hue = "Ship Mode", palette = "Set2", edgecolor = "black")
plt.title("Total Sales per Customer Segment and Ship Mode", fontsize = 15)
plt.xlabel("Customer Segment", fontsize = 14)
plt.ylabel("Total Sales", fontsize = 14)
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# 4) Which product category has the highest profit-to-sales ratio?

category_summary = a.groupby("Product Category")[["Sales", "Profit"]].sum()
category_summary["Profit_to_Sales_Ratio"] = category_summary["Profit"] / category_summary["Sales"]
category_summary = category_summary.sort_values("Profit_to_Sales_Ratio", ascending = False)
plt.figure(figsize = (12, 6))
sns.barplot(
    x = category_summary.index,
    y = category_summary["Profit_to_Sales_Ratio"],
    palette = "Set2",
    edgecolor = "black"
)
plt.title("Profit-to-Sales Ratio by Product Category", fontsize = 15)
plt.xlabel("Product Category", fontsize = 14)
plt.ylabel("Profit to Sales Ratio", fontsize = 14)
plt.tight_layout()
plt.show()
top_category = category_summary["Profit_to_Sales_Ratio"].idxmax()
print(f"The product category with the highest profit-to-sales ratio is: {top_category}")
```

```python
# 5) Find top 10 customers who generated the most profit?

top_customers = a.groupby("Customer Name")["Profit"].sum().nlargest(10).sort_values()
plt.figure(figsize = (12, 6))
sns.barplot(x = top_customers.values, y = top_customers.index, palette = "crest", edgecolor = "black")
plt.title("Top 10 Customers by Total Profit", fontsize = 15)
plt.xlabel("Total Profit", fontsize = 14)
plt.ylabel("Customer Name", fontsize = 14)
plt.tight_layout()
plt.show()
```

```python
# 6) Analyze monthly sales trends across years?

a["Order Date"] = pd.to_datetime(a["Order Date"])
a["Year"] = a["Order Date"].dt.year
a["Month"] = a["Order Date"].dt.month_name()
a["Month"] = pd.Categorical(a["Month"],
                           categories = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "Octomber", "November",
                                        "December"],
                           ordered = True)
monthly_sales = a.groupby(["Year", "Month"])["Sales"].sum().reset_index()
sales_pivot = monthly_sales.pivot(index = "Month", columns = "Year", values = "Sales")
plt.figure(figsize = (12, 6))
sales_pivot.plot(marker = "o", linewidth = 2)
plt.title("Monthly Sales Trends Across Years", fontsize = 15)
plt.xlabel("Month", fontsize = 14)
plt.ylabel("Total Sales", fontsize = 14)
plt.grid(True)
plt.xticks(rotation = 45)
plt.tight_layout()
plt.legend(title = "Year")
plt.show()
```

```python
# 7) Plot the profit over time (Order Date)?

a["Order Date"] = pd.to_datetime(a["Order Date"])
a = a.sort_values("Order Date")
a["Profit_Rolling"] = a["Profit"].rolling(window = 30).mean()
plt.figure(figsize = (12, 6))
plt.plot(a["Order Date"], a["Profit"], alpha = 0.3, label = "Daily Profit", color = "skyblue")
plt.plot(a["Order Date"], a["Profit_Rolling"], label = "30-Day Rolling Average", color = "darkblue", linewidth = 2)
plt.title("Profit Over Time With 30-Day Rolling Average", fontsize = 15)
plt.xlabel("Order Date", fontsize = 14)
plt.ylabel("Profit", fontsize = 14)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 8) Find the month with the highest number of orders for each year?

a["Order Date"] = pd.to_datetime(a["Order Date"])
a["Year"] = a["Order Date"].dt.year
a["Month"] = a["Order Date"].dt.month
a["Month_Name"] = a["Order Date"].dt.month_name()
monthly_orders = a.groupby(["Year", "Month", "Month_Name"]).size().reset_index(name = "Order_Count")
max_orders_per_year = monthly_orders.loc[monthly_orders.groupby("Year")["Order_Count"].idxmax()]
print("Month with the highest number of orders for each year:")
print(max_orders_per_year[["Year", "Month_Name", "Order_Count"]])
```

```python
# 9) Compare total sales in Q1, Q2, Q3, Q4?

a["Order Date"] = pd.to_datetime(a["Order Date"])
a["Quarter"] = a["Order Date"].dt.quarter
quarterly_sales = a.groupby("Quarter")["Sales"].sum().reset_index()
quarter_labels = {1: "Q1", 2: "Q2", 3: "Q3", 4: "Q4"}
quarterly_sales["Quarter"] = quarterly_sales["Quarter"].map(quarter_labels)
plt.figure(figsize=(12, 6))
plt.bar(quarterly_sales["Quarter"], quarterly_sales["Sales"], color = 'skyblue', edgecolor = "black")
plt.title("Total Sales by Quarter", fontsize = 15)
plt.xlabel("Quarter", fontsize = 14)
plt.ylabel("Sales", fontsize = 14)
plt.grid(axis = "y")
plt.tight_layout()
plt.show()
plt.figure(figsize = (12, 6))
plt.pie(quarterly_sales["Sales"], labels = quarterly_sales["Quarter"], autopct = "%1.1f%%",
        startangle = 90, colors = ["#66b3ff", "#99ff99", "#ffcc99", "#ff9999"])
plt.title("Quarterly Sales Distribution", fontsize = 15)
plt.tight_layout()
plt.show()
```

```python
# 10) Plot a boxplot of shipping cost across different product categories.

plt.figure(figsize = (12, 6))
sns.boxplot(data = a, x = "Product Category", y = "Shipping Cost", palette = "Pastel1")
plt.title("Shipping Cost Distribution by Product Category", fontsize = 15)
plt.xlabel("Product Category", fontsize = 14)
plt.ylabel("Shipping Cost", fontsize = 14)
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# 11) Create a violin plot of profit margins by product category.

a["Profit Margin"] = a["Profit"] / a["Sales"]
plt.figure(figsize = (12, 6))
sns.violinplot(data = a, x = "Product Category", y = "Profit Margin", palette = "Set2")
plt.title("Profit Margins by Product Category", fontsize = 15)
plt.xlabel("Product Category", fontsize = 14)
plt.ylabel("Profit Margin", fontsize = 14)
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# 12) Draw a scatter plot of Unit Price vs Profit, colored by Region.

plt.figure(figsize = (12, 6))
sns.scatterplot(data = a, x = "Unit Price", y = "Profit", hue = "Region", palette = "Set2")
plt.title("Unit Price vs Profit (Colored by Region)", fontsize = 15)
plt.xlabel("Unit Price", fontsize = 14)
plt.ylabel("Profit", fontsize = 14)
plt.grid()
plt.legend(title = "Region", bbox_to_anchor = (1.05, 1), loc = "upper left")
plt.tight_layout()
plt.show()
```

```python
# 13) Make a stacked bar chart of Quantity ordered per Region and Customer Segment.

quantity_data = a.groupby(["Region", "Customer Segment"])["Quantity ordered new"].sum().unstack()
quantity_data.plot(kind = "bar", stacked = True, figsize = (12, 6), colormap = "Set3", edgecolor = "black")
plt.title("Quantity Ordered Per Region and Customer Segment", fontsize = 15)
plt.xlabel("Region", fontsize = 14)
plt.ylabel("Total Quantity Ordered", fontsize = 14)
plt.legend(title = "Customer Segment", bbox_to_anchor = (1.05, 1), loc = "upper left")
plt.tight_layout()
plt.show()
```

```python
# 14) Use a heatmap to visualize correlation among numeric columns (Sales, Profit, Discount, Unit Price).

numeric_cols = a[["Sales", "Profit", "Discount", "Unit Price"]]
corr_matrix = numeric_cols.corr()
plt.figure(figsize = (12, 6))
sns.heatmap(corr_matrix, annot = True, cmap = "YlOrBr", fmt = ".2f", linewidth = 0.5)
plt.title("Correlation Heatmap - Sales, Profit, Discount, Unit Price", fontsize = 15)
plt.tight_layout()
plt.show()
```

```python
# 15) Find orders where discount > 20% and profit < 0. Investigate which categories are impacted.

high_discount_loss = a[(a["Discount"] > 0.20) & (a["Profit"] < 0)]
category_impact = high_discount_loss["Product Category"].value_counts()
plt.figure(figsize = (12, 6))
category_impact.plot(kind = "bar", color = "tomato", edgecolor = "black")
plt.title("Category With Lost Due To High Discount (>20%)", fontsize = 15)
plt.xlabel("Product Category", fontsize = 14)
plt.ylabel("Number of Loss-Making Orders", fontsize = 14)
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# 16) Create a new column called Profit Margin (%) = Profit / Sales * 100. Find the top 5 products by this margin.

a["Profit Margin (%)"] = (a["Profit"] / a["Sales"])*100
top_5_margin_products = a[["Product Name", "Profit Margin (%)"]].sort_values(by = "Profit Margin (%)", ascending = False).dropna().head(5)
print(top_5_margin_products)
```

```python
# 17) Find the average delivery time (Ship Date - Order Date) per region. Sort in descending order.
a["Order Date"] = pd.to_datetime(a["Order Date"])
a["Ship Date"] = pd.to_datetime(a["Ship Date"])
a["Delivery Time (Days)"] = (a["Ship Date"] - a["Order Date"]).dt.days
avg_delivery_by_region = a.groupby("Region")["Delivery Time (Days)"].mean().sort_values(ascending = False)
print(avg_delivery_by_region)
```

```python
# 18) Which product sub-categories often have high discount and low profit? (avg discount > 15% and avg profit < 0)

subcat_analysis = a.groupby("Product Sub-Category")[["Discount", "Profit"]].mean()
high_discount_low_profit = subcat_analysis[(subcat_analysis["Discount"]>0.15) & (subcat_analysis["Profit"]<0)]
print(high_discount_low_profit)
```

```python
# 19) Which products are frequently sold together? (Market Basket style using Order ID)

from itertools import combinations
from collections import Counter
order_product = a.groupby("Order ID")["Product Name"].apply(list)
product_pairs = Counter()
for products in order_product:
    pairs = combinations(sorted(set(products)), 2)
    product_pairs.update(pairs)
    top_pairs = product_pairs.most_common(10)
for pair, count in top_pairs:
    print(f"{pair[0]} & {pair[1]}----{count} times")
```

```python
# 20) What are the top 5 most ordered products in each region?

product_orders = a.groupby(["Region", "Product Name"])["Quantity ordered new"].sum().reset_index()
top5_per_region = product_orders.groupby("Region").apply(
    lambda x:x.nlargest(5, "Quantity ordered new")
).reset_index(drop = True)
print(top5_per_region)
```

```python
# 21) Which customers have the highest return on investment (ROI = Profit / Quantity)?

a["ROI"] = a["Profit"] / a["Quantity ordered new"]
roi_per_customer = a.groupby("Customer Name")["ROI"].mean().reset_index()
top_roi_customers = roi_per_customer.sort_values(by = "ROI", ascending = False).head(10)
print(top_roi_customers)
```

```python
# 22) Create a pivot table showing total Sales and Profit per Product Category and Region.

pivot_sales_profit = a.pivot_table(
    values = ["Sales", "Profit"],
    index = "Product Category",
    columns = "Region",
    aggfunc = "sum",
    margins = True,
    margins_name = "Total"
)
print(pivot_sales_profit)
```

```python
# 23) Segment customers using KMeans clustering based on their sales and profit behavior.

from sklearn.cluster import KMeans
customer_data = a.groupby("Customer ID")[["Sales", "Profit"]].sum().reset_index()
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(customer_data[["Sales", "Profit"]])
kmeans = KMeans(n_clusters = 4, random_state = 42)
customer_data["Cluster"] = kmeans.fit_predict(scaled_data)
plt.figure(figsize = (10, 6))
sns.scatterplot(data = customer_data, x = "Sales", y = "Profit", hue = "Cluster", palette = "Set2", s = 100)
plt.title("Customer Segmentation Based on Sales and Profit", fontsize = 15)
plt.xlabel("Total Sales", fontsize = 14)
plt.ylabel("Total Profit", fontsize = 14)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 24) Forecast sales using Prophet or ARIMA on the Order Date and Sales.

from prophet import Prophet
df = a[["Order Date", "Sales"]].copy()
df = df.groupby("Order Date").sum().reset_index()
df.rename(columns = {"Order Date": "ds", "Sales": "y"}, inplace = True)
model = Prophet()
model.fit(df)
future = model.make_future_dataframe(periods = 90)
forecast = model.predict(future)
fig1 = model.plot(forecast)
plt.title("Sales Forecast using Prophet", fontsize = 15)
plt.xlabel("Date", fontsize = 14)
plt.ylabel("Sales", fontsize = 14)
plt.tight_layout()
plt.show()
fig2 = model.plot_components(forecast)
plt.tight_layout()
plt.show()
```
